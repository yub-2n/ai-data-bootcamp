```python
import requests
from bs4 import BeautifulSoup
```

```python
url = 'https://news.naver.com/section/103'
response = requests.get(url)

print('상태 코드:', response.status_code)    # 200 = 정상
print('인코딩:   ', response.encoding)
print('HTML 길이:', len(response.text), '자')
print()
print('HTML 앞부분 미리보기:')
print(response.text[:200])
```

```python
soup = BeautifulSoup(response.text, "html.parser")

selector ='ul[id^="_SECTION_HEADLINE_LIST"]'


headlines =soup.select_one(selector)
for i, li in enumerate(headlines.select("li"), 1):
    title_el = li.select_one("strong.sa_text_strong")
    press_el = li.select_one("div.sa_text_press")
    if title_el and press_el:
        print(f"{i}. {title_el.get_text(strip=True)}, {press_el.get_text(strip=True)}")


```
